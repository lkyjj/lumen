# LUMEN Quickstart
离线复现配置校验、完整 DAG、预算与一句话标准测试；默认不会联网或产生费用。

```python
from pathlib import Path

from lumen.config import load_project
from lumen.one_sentence import plan_one_sentence
from lumen.orchestrator import build_dag, render_dry_run

film = Path('../projects/vanishing-light/film.yaml').resolve()
project = load_project(film)
print(render_dry_run(project, build_dag(project, film.parent)))
```

```python
plan_one_sentence('最后一座灯塔熄灭前，守塔人看见海面升起第二个太阳。', film=film)
```
